## Join & QA the Data

```python
import pandas as pd
```

```python
churn_customer_info = pd.read_excel('Bank_Churn_Messy.xlsx')
```

```python
churn_customer_info.head()
```

```python
churn_account_info = pd.read_excel('Bank_Churn_Messy.xlsx', sheet_name=1)
```

```python
churn_account_info.head()
```

```python
churn_df=churn_customer_info.merge(churn_account_info, how="left", on="CustomerId")
```

```python
churn_df.head()
```

```python
churn_df.duplicated(keep=False)
```

```python
churn_df = churn_df.drop('Tenure_y',axis=1).rename({'Tenure_x':'Tenure'},axis=1).drop_duplicates()
```

```python
churn_df.info()
```

```python
churn_df.head()
```

```python
churn_df['EstimatedSalary'] = churn_df['EstimatedSalary'].str.replace('€','').astype('float')
churn_df['Balance'] = churn_df['Balance'].str.replace('€','').astype('float')
```

```python
churn_df.info()
```

## Cleaning The Data

```python
# checking those rows where is a missing values placed
churn_df[churn_df.isna().any(axis=1)==True]
```

```python
churn_df = churn_df.fillna(value={'Surname':'Missing', 'Age': churn_df['Age'].median()})
```

```python
churn_df.info()
```

```python
churn_df.iloc[[28,121,9389]]
```

```python
churn_df.describe()
```

```python
churn_df['EstimatedSalary']= churn_df['EstimatedSalary'].replace(-999999,churn_df['EstimatedSalary'].median())
```

```python
churn_df.describe()
```

```python
churn_df['Geography'].value_counts()
```

```python
import numpy as np
churn_df['Geography']=np.where(churn_df['Geography'].isin(['FRA','French','France']),'France',churn_df['Geography'])
```

```python
churn_df['Geography'].value_counts()
```

## Feature Engineering & Data Prepare

```python
churn_df.head()
```

```python
modelling_df = churn_df.drop(['CustomerId','Surname'],axis=1)
```

```python
modelling_df.head()
```

```python
modelling_df=pd.get_dummies(modelling_df,drop_first=True)
```

```python
modelling_df['balance_v_income']=modelling_df['Balance']/modelling_df['EstimatedSalary']
```

```python
modelling_df.head()
```

```python
modelling_df.describe()
```

```python
sns.boxplot(data=modelling_df.query('balance_v_income < 10'),y='balance_v_income')
```
